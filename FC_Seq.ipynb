```python
from Bio import SeqIO

input_fasta = 'data/secuencias.fasta.txt'
clean_fasta = 'data/secuencias_clean.fasta'

with open(clean_fasta, 'w') as out_handle:
    for record in SeqIO.parse(input_fasta, 'fasta'):
        seq = str(record.seq).upper().replace('X', 'N')
        record.seq = seq
        SeqIO.write(record, out_handle, 'fasta')

print(f"Secuencias limpias guardadas en: {clean_fasta}")
```

```python
from Bio import SeqIO
from Bio.Align.Applications import ClustalOmegaCommandline
import os

# Archivos de entrada
transcritos_path = 'data/transcritos_nuevos.fasta'
blast_hits_path = 'results/FC_results/seqdump.txt'
archivo_combinado = 'results/FC_results/combinado.fasta'
alineamiento_output = 'results/FC_results/alineamiento.aln'

# Combinar FASTA original + secuencias BLAST
registros_transcritos = list(SeqIO.parse(transcritos_path, "fasta"))
registros_blast = list(SeqIO.parse(blast_hits_path, "fasta"))

# Guardar combinado
with open(archivo_combinado, "w") as out_fasta:
    SeqIO.write(registros_transcritos + registros_blast, out_fasta, "fasta")

# Ejecutar Clustal Omega (asegúrate de tenerlo instalado)
clustalomega_cline = ClustalOmegaCommandline(
    infile=archivo_combinado,
    outfile=alineamiento_output,
    outfmt="clu",  # formato Clustal
    force=True
)
stdout, stderr = clustalomega_cline()

print("✅ Alineamiento completado. Guardado en:", alineamiento_output)
```
