#PREPROCESSING - LIMPIEZA DE SECUENCIAS

```python
#UN ARCHIVO CON TODAS LAS SECUENCIAS LIMPIADAS: cleaned_sequences

import re
import os

def limpiar_secuencias(input_file, output_file):
    """Limpia un archivo FASTA, normaliza headers y cambia formato de nombres"""
    try:
        with open(input_file, 'r') as f_in, open(output_file, 'w') as f_out:
            for line in f_in:
                if line.startswith('>'):  # Si es cabecera
                    # Normalizar header y cambiar formato
                    cleaned_header = re.sub(r'^>\s*unknown sequence (\d+)', r'>\1 unknown sequence \1', line)
                    f_out.write(cleaned_header)
                else:  # Si es secuencia
                    # Eliminar x/X/n/N y caracteres no deseados
                    cleaned_line = re.sub(r'[xXnN]', '', line)
                    f_out.write(cleaned_line)
        return True
    except Exception as e:
        print(f"Error: {str(e)}")
        return False

# Configuración de rutas
input_path = r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\data\secuencias.fasta.txt"
output_dir = r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\identificacion_patogeno\identificacion_patogeno_2\cleaned_sequences"
output_filename = "cleaned_sequences.fasta"
output_path = os.path.join(output_dir, output_filename)

# Asegurar que el directorio existe
os.makedirs(output_dir, exist_ok=True)

# Ejecutar
if limpiar_secuencias(input_path, output_path):
    print(f"¡Éxito! Archivo creado en: {output_path}")
    print("Contenido del directorio de salida:")
    print(os.listdir(output_dir))
    
    # Verificación adicional
    print("\nVerificación de headers modificados:")
    with open(output_path, 'r') as f_check:
        for line in f_check:
            if line.startswith('>'):
                print(f"Header encontrado: {line.strip()}")
else:
    print("No se pudo completar la operación")
```

```python
#ARCHIVOS PARA CADA SECUENCIA LIMPIADA POR SEPARADO: individual_sequences

import re
import os

def limpiar_secuencias(archivo_entrada, directorio_salida):
    """Procesa un archivo FASTA creando archivos individuales con nuevos nombres"""
    
    # Crear directorio si no existe
    os.makedirs(directorio_salida, exist_ok=True)
    
    with open(archivo_entrada, 'r') as f:
        contenido = f.read()
    
    # Expresión regular mejorada para manejar espacios después de >
    secuencias = re.split(r'(>\s*unknown sequence \d+)', contenido)
    secuencias = [s for s in secuencias if s.strip()]  # Eliminar cadenas vacías
    
    archivos_creados = 0
    
    # Procesar en pares (header, secuencia)
    for i in range(0, len(secuencias), 2):
        if i+1 >= len(secuencias):
            break
            
        header = secuencias[i].strip()
        secuencia = secuencias[i+1].strip()
        
        # Extraer número de secuencia y crear nuevo header
        try:
            num_secuencia = re.search(r'(\d+)$', header).group(1)
            nuevo_header = f">{num_secuencia} unknown sequence {num_secuencia}"
        except:
            continue
        
        # Limpiar la secuencia
        secuencia_limpia = re.sub(r'[xXnN\s]', '', secuencia)
        
        # Crear archivo con nuevo nombre
        nombre_archivo = f"sequence_{num_secuencia}.fasta"
        ruta_completa = os.path.join(directorio_salida, nombre_archivo)
        
        with open(ruta_completa, 'w') as f_out:
            f_out.write(f"{nuevo_header}\n{secuencia_limpia}\n")
            archivos_creados += 1
    
    return archivos_creados

# Configuración
archivo_input = r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\data\secuencias.fasta.txt"
directorio_output = r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\identificacion_patogeno\identificacion_patogeno_2\individual_sequences"

# Ejecutar
try:
    num_archivos = limpiar_secuencias(archivo_input, directorio_output)
    print(f"¡Éxito! Se crearon {num_archivos} archivos en {directorio_output}")
    
    # Verificación detallada
    print("\nVerificación de archivos (1-13):")
    for i in range(1, 14):
        archivo = f"sequence_{i}.fasta"
        ruta = os.path.join(directorio_output, archivo)
        if os.path.exists(ruta):
            with open(ruta, 'r') as f:
                lineas = f.readlines()
                print(f"- {archivo} ✓ (Header: {lineas[0].strip()})")
        else:
            print(f"- {archivo} ✗ (no encontrado)")
    
except Exception as e:
    print(f"Error: {str(e)}")
    print("\nRecomendaciones:")
    print("1. Verifica manualmente el archivo de entrada")
    print("2. Revisa los permisos de escritura")
    print("3. Comprueba que no haya caracteres especiales")
```
