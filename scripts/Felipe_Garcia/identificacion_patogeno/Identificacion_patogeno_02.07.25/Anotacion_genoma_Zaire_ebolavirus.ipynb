```python
from Bio import SeqIO
from pathlib import Path

# === RUTAS ===
gbff_file = Path(r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\data\Identificacion_patogeno\Zaire_ebolavirus_genome\ncbi_dataset\data\GCA_000848505.1\genomic.gbff")
output_folder = Path(r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\identificacion_patogeno\identificacion_patogeno_4_02.07.25\Anotacion_genoma_Zaire_ebolavirus")
output_folder.mkdir(parents=True, exist_ok=True)

output_table = output_folder / "tabla_anotacion_genoma.tsv"

# === EXTRACCIÓN ===
with open(output_table, "w", encoding="utf-8") as out:
    out.write("Gene\tStart\tEnd\tStrand\tProduct\n")
    for record in SeqIO.parse(gbff_file, "genbank"):
        for feature in record.features:
            if feature.type == "CDS":
                gene = feature.qualifiers.get("gene", ["-"])[0]
                product = feature.qualifiers.get("product", ["-"])[0]
                start = int(feature.location.start) + 1  # posiciones 1-based
                end = int(feature.location.end)
                strand = "+" if feature.location.strand == 1 else "-"
                out.write(f"{gene}\t{start}\t{end}\t{strand}\t{product}\n")

print(f"✅ Tabla generada: {output_table}")
```

```python
from Bio import SeqIO
from dna_features_viewer import GraphicFeature, GraphicRecord
import matplotlib.pyplot as plt
from pathlib import Path

# Rutas
gbff_path = Path(r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\data\Identificacion_patogeno\Zaire_ebolavirus_genome\ncbi_dataset\data\GCA_000848505.1\genomic.gbff")
output_img = Path(r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\identificacion_patogeno\identificacion_patogeno_4_02.07.25\Anotacion_genoma_Zaire_ebolavirus\mapa_genomico.png")

# Extraer features
features = []

for record in SeqIO.parse(gbff_path, "genbank"):
    sequence_length = len(record.seq)
    for feature in record.features:
        if feature.type == "CDS":
            start = int(feature.location.start)
            end = int(feature.location.end)
            strand = +1 if feature.location.strand == 1 else -1
            gene = feature.qualifiers.get("gene", [""])[0]
            product = feature.qualifiers.get("product", [""])[0]

            label = f"{gene} ({product})" if gene else product

            features.append(GraphicFeature(start=start,
                                           end=end,
                                           strand=strand,
                                           color="#ffd700" if strand == 1 else "#87cefa",
                                           label=label))

# Crear gráfico
record = GraphicRecord(sequence_length=sequence_length, features=features)
ax, _ = record.plot(figure_width=12)

# Guardar imagen
plt.tight_layout()
plt.savefig(output_img, dpi=300)
plt.close()

print(f"✅ Mapa genómico guardado en: {output_img}")
```
