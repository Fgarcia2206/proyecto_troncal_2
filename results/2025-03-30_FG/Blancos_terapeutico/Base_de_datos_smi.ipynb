```python
import pandas as pd

input_path = r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\Blancos_terapeutico\ncats_target_based_curated.xlsx"
output_path = r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\Blancos_terapeutico\smacc_all_compounds.smi"

# Cargar la hoja correcta (con espacio entre all y data)
df = pd.read_excel(input_path, sheet_name="master_table (all data points)")

# Limpiar espacios en nombres de columnas
df.columns = df.columns.str.strip()

# Filtrar filas con SMILES y Molecule ChEMBL ID no nulos y no vacíos
df = df[df['SMILES'].notna() & (df['SMILES'] != '')]
df = df[df['Molecule ChEMBL ID'].notna() & (df['Molecule ChEMBL ID'] != '')]

# Eliminar duplicados según la columna SMILES
df = df.drop_duplicates(subset=['SMILES'])

# Guardar archivo .smi
with open(output_path, "w") as f:
    for _, row in df.iterrows():
        f.write(f"{row['SMILES']}\t{row['Molecule ChEMBL ID']}\n")

```
