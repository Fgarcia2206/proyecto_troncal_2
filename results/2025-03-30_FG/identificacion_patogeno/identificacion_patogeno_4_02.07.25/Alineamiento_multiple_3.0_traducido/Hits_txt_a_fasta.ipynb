```python
import re
from Bio import Entrez
from pathlib import Path

Entrez.email = "tu_correo@ejemplo.com"

input_folder = Path(r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\identificacion_patogeno\identificacion_patogeno_4_02.07.25\Alineamiento_multiple_3.0_traducido\Blastn_query_separados")
output_folder = Path(r"C:\Users\fgarc\OneDrive\Escritorio\Doctorado\Ramos\1° Semestre\Troncal\proyecto_troncal2\results\2025-03-30_FG\identificacion_patogeno\identificacion_patogeno_4_02.07.25\Alineamiento_multiple_3.0_traducido\Hits_a_fasta")
output_folder.mkdir(exist_ok=True)

# Regex para accesiones tipo WWV92822.1, Q2PDK5.1, etc.
accession_pattern = re.compile(r'\b([A-Z]{3,4}\d{5,6}\.\d)\b')

for txt_file in input_folder.glob("*.txt"):
    ids = set()
    with open(txt_file, encoding="utf-8") as f:
        for line in f:
            matches = accession_pattern.findall(line)
            ids.update(matches)

    if not ids:
        print(f"{txt_file.name} → No hay accesiones válidas encontradas")
        continue

    print(f"{txt_file.name} → Accesiones válidas: {list(ids)}")

    try:
        handle = Entrez.efetch(db="protein", id=",".join(ids), rettype="fasta", retmode="text")
        output_path = output_folder / f"{txt_file.stem}.fasta"
        with open(output_path, "w") as out_f:
            out_f.write(handle.read())
        print(f"Archivo FASTA generado: {output_path}")
    except Exception as e:
        print(f"Error con {txt_file.name}: {e}")

```
