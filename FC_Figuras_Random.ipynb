```python
import geopandas as gpd
import matplotlib.pyplot as plt

# Lista con solo Rusia
countries_to_highlight = ['Russia']

# Cargar shapefile de países
shapefile_path = 'mapas/mapa2/ne_110m_admin_0_sovereignty.shp'
world = gpd.read_file(shapefile_path)

# Filtrar Rusia
highlighted_countries = world[world['NAME'].isin(countries_to_highlight)]

# Graficar
fig, ax = plt.subplots(figsize=(10, 5))
world.plot(ax=ax, color='lightgrey', edgecolor='white')  # Mapa base
highlighted_countries.plot(ax=ax, color='black')  # Solo Rusia resaltada

# Quitar ejes
ax.axis('off')

# Título (vacío en este caso)
ax.set_title('', fontsize=15)

plt.tight_layout()
plt.show()
```
