```python
from Bio import SeqIO

input_fasta = 'data/secuencias.fasta.txt'

# Leer todas las secuencias del archivo FASTA
secuencias = list(SeqIO.parse(input_fasta, "fasta"))

# Imprimir información básica de cada secuencia
for secuencia in secuencias:
    print(f"ID: {secuencia.id}")
    print(f"Descripción: {secuencia.description}")
    print(f"Secuencia (primeros 50 nt): {secuencia.seq[:50]}")
    print(f"Longitud: {len(secuencia.seq)} nucleótidos")
    print("-" * 40)
```
